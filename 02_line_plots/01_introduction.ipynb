***Line plots are one of the most commonly used visualizations in Data Science.***

````console
When do we use line plots?

Whenever we want to understand how something changes over time or across an ordered sequence.

Examples:
- Monthly sales trends.
- Daily temperatures.
- Stock prices over time.
- Website traffic.
- Student performance over semesters.
````

`````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [10000, 15000, 12000, 20000, 25000]

ax.plot(months, sales)

plt.show()
`````

## Changing line colors

`````python
fig, ax = plt.subplots()

ax.plot(months, sales, color = "red")

plt.show()
`````

## Changing line thickness

* linewidth=1 → Thin line.
* linewidth=2 → Medium line.
* linewidth=3 → Thick line.

`````python
fig, ax = plt.subplots()

ax.plot(months, sales, color = "red", linewidth = 3)

plt.show()
`````

## Changing line styles

```console
Common line styles 

Solid -
Dashed --
Dotted :
Dash-dot -.
```

`````python
fig, ax = plt.subplots()

ax.plot(months, sales, linestyle="--")

plt.show()
`````

## Adding markers and Changing marker size

`````python
fig, ax = plt.subplots()

ax.plot(months, sales, marker="o", markersize = 10)

plt.show()
`````

## Combining everything

`````python
fig, ax = plt.subplots()

ax.set_title("Monthly Sales")

ax.set_xlabel("Months")

ax.set_ylabel("Sales")

ax.plot(months, sales, color = "green", linewidth = 2, marker = 'o', markersize = 8, linestyle = '-.')

plt.show()
`````

## Plotting multiple lines

```console
New concept: label and legend()

- label gives a name to each line.
- ax.legend() displays those names on the graph.

Without ax.legend(), the labels won't appear.
```

`````python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

product_A = [10000, 15000, 12000, 20000, 25000]
product_B = [8000, 12000, 16000, 18000, 22000]

fig, ax = plt.subplots()

ax.set_title("Product Sales Comparison")
ax.set_xlabel("Months")
ax.set_ylabel("Sales")

ax.plot(months, product_A, label = "Product A", marker = 'o')
ax.plot(months, product_B, label = "Product B", marker = 's')

ax.legend() # gives legend

plt.show()
`````
