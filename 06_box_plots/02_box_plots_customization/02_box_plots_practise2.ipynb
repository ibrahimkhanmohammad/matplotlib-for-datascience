## Task 1 — Customize the salary box plot

Use the previous salary dataset.

Requirements:
1. Create a box plot.
2. Enable patch_artist.
3. Display the mean.
4. Change the box color to light green.
5. Set an appropriate title and Y-axis label.

```python
import matplotlib.pyplot as plt

salaries = [
    20000, 22000, 25000, 27000, 30000,
    32000, 35000, 38000, 40000, 42000,
    45000, 48000, 50000, 55000, 100000
]

fig, ax = plt.subplots()

box = ax.boxplot(salaries, patch_artist = True, showmeans = True)

box["boxes"][0].set_facecolor("lightgreen")

ax.set_title("Employee Salary Distribution")
ax.set_ylabel("Salary")

plt.show()
```

## Task 2 — Department comparison
    use:
    it = [25000, 30000, 35000, 40000, 45000]
    hr = [20000, 22000, 25000, 27000, 30000]
    sales = [18000, 25000, 30000, 40000, 60000]

Requirements:
1. Create one box plot containing all three departments.
2. Label the boxes IT, HR and Sales.
3. Add a title and Y-axis label.
4. Display the plot.

```python
import matplotlib.pyplot as plt

it = [25000, 30000, 35000, 40000, 45000]
hr = [20000, 22000, 25000, 27000, 30000]
sales = [18000, 25000, 30000, 40000, 60000]

fig, ax = plt.subplots()

ax.boxplot([it, hr, sales])

ax.set_xticks([1, 2, 3])
ax.set_xticklabels(["IT", "HR", "Sales"])

ax.set_title("Department wise Salary distribution")
ax.set_ylabel("Salary")

plt.show()
```

## Task 3 — Conceptual review

1. What does patch_artist=True do?

It allows us to fill the box with a color.
Without it, the box is generally drawn without a filled face.


2. What does showmeans=True display?

It displays the mean (average) of the dataset as a marker.

Remember:
* Median = middle value.
* Mean = arithmetic average.
They are not necessarily the same.


3. Vertical vs horizontal box plots

* Vertical box plot → Values are represented along the Y-axis.
* Horizontal box plot → Values are represented along the X-axis.

The horizontal version is especially useful when comparing categories with long names.


4. Why compare multiple departments using box plots?

* Comparing department medians.
* Comparing salary distributions.
* Identifying potential outliers.
* Comparing multiple groups in a single visualization.

That's exactly how box plots are used in EDA.

One distinction: a box plot doesn't directly tell us the highest individual salary if that value isn't shown as a whisker or outlier; we can inspect the underlying data for the exact maximum.


5. What does df.values do?

* df.values → Returns the underlying data as a NumPy array.
* df.columns → Returns column names.
* df.index → Returns row index labels.
