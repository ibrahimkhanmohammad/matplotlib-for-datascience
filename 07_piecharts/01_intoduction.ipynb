## What is a pie chart?
***A pie chart represents how a whole is divided into different categories.***

For example, imagine a student's weekly study time:
* Python: 10 hours
* Mathematics: 5 hours
* Statistics: 3 hours
* SQL: 2 hours

A pie chart helps us understand what proportion of the total study time each subject occupies.

## Basic Syntax

Understand these arguments:
* hours — numerical values that determine each slice's size.
* labels — names displayed beside the slices.
* ax.pie() — creates the pie chart.

```python
import matplotlib.pyplot as plt

subjects = ["Python", "Maths", "Statistics", "SQL"]
hours = [10, 5, 3, 2]

fig, ax = plt.subplots()

ax.pie(hours, labels=subjects)

ax.set_title("Weekly Study Time")

plt.show()
```

## Important customization parameters

### autopct — display percentages

* %1.1f%% displays percentages with one decimal place.
* Example: 50.0%, 25.0%.

```python
fig, ax = plt.subplots()

ax.pie(hours, labels=subjects, autopct="%1.1f%%")

plt.show()
```

### startangle — rotate the pie chart
***Rotates the starting position of the pie chart.***

```python
fig, ax = plt.subplots()

ax.pie(hours, labels=subjects, autopct="%1.1f%%", startangle = 90)

plt.show()
```

### explode — separate a slice

***Here, the first slice (Python) moves slightly away from the center.***

```python
fig, ax = plt.subplots()

explode = [0.1, 0, 0, 0]

ax.pie(
    hours,
    labels=subjects,
    autopct="%1.1f%%",
    startangle = 90,
    explode=explode
)

plt.show()
```

### colors — customize slice colors

```python
fig, ax = plt.subplots()

colors = ["skyblue", "orange", "lightgreen", "pink"]

ax.pie(
    hours,
    labels=subjects,
    autopct="%1.1f%%",
    startangle = 90,
    colors = colors
)

plt.show()
```

## Complete example

```python
import matplotlib.pyplot as plt

subjects = ["Python", "Maths", "Statistics", "SQL"]
hours = [10, 5, 3, 2]

fig, ax = plt.subplots()

explode = [0.1, 0, 0, 0]
colors =  ["skyblue", "pink", "lightgreen", "orange"]


ax.pie(hours, labels = subjects, autopct = "%1.1f%%", startangle = 90, explode = explode, colors = colors)

ax.set_title("Weekly Study Time")

plt.show()
```

## When should you use pie charts?
Use them when:
- You want to show proportions of a whole.
- You have a small number of categories.
- All values are non-negative and represent parts of the same total.
Avoid them when:
- You have too many categories.
- You need precise comparisons between similar values.
- The values do not represent parts of a meaningful whole.
Data Science tip: Bar charts are often better for comparing categories. Pie charts are useful, but don't overuse them.
