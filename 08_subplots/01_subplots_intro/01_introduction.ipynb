## What are subplots?
Subplots allow us to display multiple charts inside a single figure.

For example, suppose we have student marks and want to visualize them using:
- Line plot
- Bar chart
- Scatter plot
- Histogram
Instead of opening four separate figures, we can display all four in one figure.

## Understanding plt.subplots()
You already know this:          
fig, ax = plt.subplots()
This creates:
- fig → the entire figure.
- ax → one plotting area (Axes).

Now let's create multiple plotting areas.           
fig, ax = plt.subplots(2, 2) 

What does 2, 2 mean?            
plt.subplots(rows, columns)         
It Creates 2 rows and 2 columns.

Important: Python indexing starts at 0, not 1.

## Your first subplot program

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2)

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

# First subplot
ax[0, 0].plot(months, sales)
ax[0, 0].set_title("Line Plot")

# Second subplot
ax[0, 1].bar(months, sales)
ax[0, 1].set_title("Bar Chart")

# Third subplot
ax[1, 0].scatter(months, sales)
ax[1, 0].set_title("Scatter Plot")

# Fourth subplot
ax[1, 1].hist(sales)
ax[1, 1].set_title("Histogram")

fig.tight_layout()

plt.show()
```

## What is fig.tight_layout()?

When multiple charts are displayed, titles and labels may overlap - Automatically adjusts spacing between subplots to reduce overlapping.

fig.tight_layout()

## Creating 1 row and 2 columns

You don't always need a 2×2 layout. 
    
Here, ax[0] is the first subplot and ax[1] is the second.

Important distinction:
- plt.subplots(2, 2) → use ax[row, column].
- plt.subplots(1, 2) → use ax[column].
