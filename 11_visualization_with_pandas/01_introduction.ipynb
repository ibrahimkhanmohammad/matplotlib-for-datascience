## Real-world DataFrame visualization
Now let's apply everything to a realistic product sales dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Monitor", "Keyboard"],
    "Units_Sold": [50, 120, 80, 40, 150],
    "Revenue": [2500000, 1800000, 960000, 800000, 150000],
    "Profit": [300000, 250000, 120000, 100000, 30000]
}

df = pd.DataFrame(data)

print(df)
```

## A. Revenue by product

```python
fig, ax = plt.subplots(figsize=(12, 7))

ax.bar(df["Product"], df["Revenue"], color="pink", edgecolor="black")
ax.set_title("Revenue by Product")
ax.set_xlabel("Product")
ax.set_ylabel("Revenue (₹)")

ax.grid(True, axis="y", alpha = 0.25)
fig.tight_layout()

plt.show()
```

## B. Units sold distribution

```python
fig, ax = plt.subplots(figsize=(12, 7))

ax.hist(df["Units_Sold"], bins = 5, color="orange", edgecolor="black")
ax.set_title("Units Sold Distributions")

ax.grid(True, axis="y", alpha = 0.25)
fig.tight_layout()

plt.show()
```

## C. Profit vs Revenue

```python
fig, ax = plt.subplots(figsize = (12,7))

ax.scatter(
    df["Revenue"],
    df["Profit"],
    color="red"
)

ax.set_title("Profit vs Revenue")
plt.show()
```
