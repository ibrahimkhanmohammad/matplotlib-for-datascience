        3. Create a box plot using ax.boxplot().
        4. Add a suitable title.
        5. Add a Y-axis label.
        6. Display the plot.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

salaries = [
    20000, 22000, 25000, 27000, 30000,
    32000, 35000, 38000, 40000, 42000,
    45000, 48000, 50000, 55000, 100000
]

ax.boxplot(salaries)

ax.set_title("Employee Salaries Distribution")

ax.set_ylabel("Salaries")

plt.show()
```

### ***Let's calculate the five-number summary and IQR ourselves***

salaries = [10, 20, 30, 40, 50, 60, 70]

1. the minimum value is 10
2. the q1 25th percentile is 20
3. the median 50th percentile is 40
4. the q3 75th percentile is 60
5. the maximum value is 70

Now calculate IQR:
    IQR = Q3 - Q1 = 60 - 20 = 40
