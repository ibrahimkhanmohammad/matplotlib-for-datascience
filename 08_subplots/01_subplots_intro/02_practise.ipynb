## Task 1 — Two charts
Use this data:    
  
    products = ["Laptop", "Mobile", "Tablet", "Watch"]          
    sales = [50, 100, 70, 40]

Requirements:
1. Create 1 row and 2 columns.
2. First subplot: Line plot.
3. Second subplot: Bar chart.
4. Give each subplot a title.
5. Add fig.tight_layout().

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2)

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [50, 100, 70, 40]

# first subplot: line plot
ax[0].plot(products, sales, marker='o')
ax[0].set_title("Line Plot")
ax[0].set_xlabel("Products")
ax[0].set_ylabel("Sales")

# second sub plot: bar chart
ax[1].bar(products, sales)
ax[1].set_title("Bar Chart")
ax[1].set_xlabel("Products")
ax[1].set_ylabel("Sales")

fig.tight_layout()

plt.show()
```

## Task 2 — Four charts
Use this data:

    students = ["Ali", "Ahmed", "Sara", "Zoya", "Omar"]
    marks = [85, 72, 90, 65, 78]

Requirements:
1. Create a 2×2 subplot layout.
2. First: Line plot.
3. Second: Bar chart.
4. Third: Scatter plot.
5. Fourth: Histogram.
6. Give all four charts appropriate titles.
7. Use fig.tight_layout().

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2)

students = ["Ali", "Ahmed", "Sara", "Zoya", "Omar"]
marks = [85, 72, 90, 65, 78]

# First subplot: Line plot
ax[0, 0].plot(students, marks, marker='o')
ax[0, 0].set_title("Line Plot")
ax[0, 0].set_xlabel("Students")
ax[0, 0].set_ylabel("Marks")

# Second subplot: Bar chart
ax[0, 1].bar(students, marks)
ax[0, 1].set_title("Bar Chart")
ax[0, 1].set_xlabel("Students")
ax[0, 1].set_ylabel("Marks")

# Third subplot: Scatter plot
ax[1, 0].scatter(students, marks, marker='x')
ax[1, 0].set_title("Scatter Plot")
ax[1, 0].set_xlabel("Students")
ax[1, 0].set_ylabel("Marks")

# Fourth subplot: Histogram
ax[1, 1].hist(marks, bins=5)
ax[1, 1].set_title("Histogram")
ax[1, 1].set_xlabel("Marks")
ax[1, 1].set_ylabel("Frequency")

fig.tight_layout()
plt.show()
```
