# Part 2: Plotting Multiple NumPy Datasets

## 1. Introduction
Suppose you want to compare three mathematical functions:

$y_1=x,\quad y_2=x^2,\quad y_3=x^3
$

Instead of creating three separate charts, we can display all three lines on the same axes.

This makes comparison easier.

In Data Science, this technique is useful for:
- Comparing multiple datasets.
- Comparing mathematical functions.
- Comparing model predictions.
- Visualizing different trends.

## 2. Creating multiple datasets using NumPy

import numpy as np

`x = np.linspace(-5, 5, 100)
`
Now calculate three different Y arrays.
```python

y1 = x
y2 = x ** 2
y3 = x ** 3
```

## 3. Plotting multiple datasets
Remember how we plotted multiple lines in Lesson 2?

We simply use multiple ax.plot() calls on the same Axes object.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5,100)
y1 = x
y2 = x ** 2
y3 = x ** 3

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x, y1, color="blue", label="y = x")
ax.plot(x, y2, color="green", label="y = x²")
ax.plot(x, y3, color="red", label="y = x³")

ax.set_title("Comparison of Mathematical Functions")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")

ax.legend()
ax.grid(True, linestyle="--", alpha=0.5)

fig.tight_layout()
plt.show()
````

## 4. Understand the important code
```console
Code	        Purpose
ax.plot(x, y1)	Plots the first dataset
ax.plot(x, y2)	Plots the second dataset
ax.plot(x, y3)	Plots the third dataset
label="..." 	Assigns a name to each line
```

`ax.legend()`	Displays the names of the lines

Remember: All three ax.plot() calls use the same ax, so all lines appear on one chart.


## 5. Important concept — Why use a legend?
Imagine you have three lines with different colors.

Without a legend, you might not know which line represents which function.

A legend identifies each dataset.
`ax.legend()`



You can also change its position:
`ax.legend(loc="upper left")`



Other common positions:
- "upper right"
- "lower left"
- "lower right"
- "best"
