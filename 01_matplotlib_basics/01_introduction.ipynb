## Importing Matplotlib and creating First Chart

* x represents the X-axis values.
* y represents the Y-axis values.
* plt.plot(x, y) creates a line graph.
* plt.show() displays the graph.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 40]

plt.plot(x, y)

plt.show()
```

## Adding title and label

*A graph without labels isn't very informative.*

so; *we use title, xlabel, ylabel such that we have a graph with a title and axis labels.*

```python
plt.plot(x, y)

plt.title("First Chart")

plt.xlabel("X Values")

plt.ylabel("Y Values")

plt.show()
```

***The recommended way to create plots is using subplots()***

```python
fig, ax = plt.subplots()

ax.plot(x, y)

ax.set_title("First Chart")

ax.set_xlabel("X Values")

ax.set_ylabel("Y Values")

plt.show()
```

***It is better to use the object-oriented approach because it gives us better control when creating multiple plots.***


Pyplot approach 

plt.plot()	                            

plt.title()
