## Customize a box plot
    We can customize box plots using parameters such as:
    Parameter	    Purpose
    vert	           Vertical or horizontal orientation (deprecated; use orientation in newer Matplotlib versions)
    patch_artist	   Allows filling the box with color
    showmeans	   Displays the mean
    showfliers	   Displays potential outliers
    notch	           Creates a notched box
    widths	           Controls box width

```python
import matplotlib.pyplot as plt

salaries = [
    20000, 22000, 25000, 27000, 30000,
    32000, 35000, 38000, 40000, 42000,
    45000, 48000, 50000, 55000, 100000
]

fig, ax = plt.subplots()

ax.boxplot(
            salaries, 
            patch_artist = True, 
            showmeans = True, 
            widths = 0.5
        )

ax.set_title("Employee Salary Distribution")
ax.set_ylabel("Salary")

plt.show()
```

What's new? 
* patch_artist=True → Enables filling the box with color.
* showmeans=True → Displays the mean as a separate marker.
* widths=0.5 → Controls the width of the box.

## Change the box color
Once patch_artist=True is enabled, we can customize the box.

Here:
* box stores the different graphical elements of the box plot.
* box["boxes"] accesses the box itself.
* [0] selects the first box.
* set_facecolor() changes its fill color.

```python
fig, ax = plt.subplots()

box = ax.boxplot(salaries, patch_artist=True)

box["boxes"][0].set_facecolor("lightblue")

ax.set_title("Employee Salary Distribution")
ax.set_ylabel("Salary")

plt.show()
```

## Horizontal box plots
So far, we've created vertical box plots.

But sometimes horizontal box plots are easier to read, especially when comparing multiple categories.

Use orientation="horizontal" in current Matplotlib versions:

Remember:
* Default orientation → Vertical.
* orientation="horizontal" → Horizontal.

```python
fig, ax = plt.subplots()

ax.boxplot(
    salaries,
    orientation="horizontal",
    patch_artist=True
)

ax.set_title("Employee Salary Distribution")
ax.set_xlabel("Salary")

plt.show()
```

## Comparing multiple categories using box plots
This is particularly useful in EDA.

Imagine we have employee salaries from three departments.

Now we can compare:
* Which department has the highest median salary?
* Which department has greater salary variation?
* Which department has potential outliers?  

This is much more useful than creating three separate charts.

```python
import matplotlib.pyplot as plt

it = [25000, 30000, 35000, 40000, 45000]
hr = [20000, 22000, 25000, 27000, 30000]
sales = [18000, 25000, 30000, 40000, 60000]

fig, ax = plt.subplots()

ax.boxplot([it, hr, sales])

ax.set_title("Salary Comparison by Department")
ax.set_ylabel("Salary")
ax.set_xticks([1, 2, 3])
ax.set_xticklabels(["IT", "HR", "Sales"])

plt.show()
```

## Box plot with Pandas DataFrame

Notice:
* df.values converts the DataFrame into a NumPy array.
* df.columns gives us the department names.
* We pass the numerical data to ax.boxplot().

This is how we'll eventually visualize real datasets

```python
import pandas as pd

import matplotlib.pyplot as plt

data = {
    "IT": [25000, 30000, 35000, 40000, 45000],
    "HR": [20000, 22000, 25000, 27000, 30000],
    "Sales": [18000, 25000, 30000, 40000, 60000]
}

df = pd.DataFrame(data)

fig, ax = plt.subplots()

ax.boxplot(df.values)

ax.set_xticks(range(1, len(df.columns) + 1))
ax.set_xticklabels(df.columns)

ax.set_title("Department Salary Comparison")
ax.set_ylabel("Salary")

plt.show()
```
