## Task 1 — Line chart

Create a line chart showing monthly Sales.

Requirements:
- Use line chart.
- Green line.
- Circle markers.
- Appropriate title and axis labels.

````python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 14000, 22000, 25000],
    "Expenses": [8000, 9000, 11000, 10000, 13000, 15000],
    "Customers": [100, 120, 150, 130, 180, 200],
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(df["Month"], df["Sales"], marker="o", color="green")
ax.grid(True, axis="y", linestyle="--", alpha=0.25)
ax.annotate(
    "Highest Sale",
    fontweight = "bold",
    fontsize = 10,
    xy=("Jun", 25000),
    xytext=("Apr", 22000),
    arrowprops=dict(arrowstyle="->", color="red"),
)

ax.set_title("Monthly Sales", fontweight ="bold")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

fig.tight_layout()
plt.show()
````

## Task 5 — Multiple line charts

Create one line chart showing both Sales and Expenses.

Requirements:
- Use y=["Sales", "Expenses"].
- Add markers.
- Add a legend.
- Add a title.

````python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 14000, 22000, 25000],
    "Expenses": [8000, 9000, 11000, 10000, 13000, 15000],
    "Customers": [100, 120, 150, 130, 180, 200],
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(df["Month"], df["Sales"], color="pink", marker = "o", markersize = 8, label = "Monthly Sales")
ax.plot(df["Month"], df["Expenses"], color="orange", marker = "s", markersize = 8, label = "Monthly Expenses")
ax.grid(True, axis="y", linestyle="--", alpha=0.25)


ax.set_title("Sales vs Expenses", fontweight ="bold")
ax.set_xlabel("Month")
ax.set_ylabel("Money in Rupees")
ax.legend(loc= "best")

fig.tight_layout()
plt.show()
````

## Task 6 — Horizontal bar chart

Create a horizontal bar chart showing monthly Sales.

Use "barh".

````python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 14000, 22000, 25000],
    "Expenses": [8000, 9000, 11000, 10000, 13000, 15000],
    "Customers": [100, 120, 150, 130, 180, 200],
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(10, 5))

ax.barh(df["Month"], df["Sales"], color="blue", alpha = 0.5)
ax.grid(True, axis="x", linestyle="--", alpha=0.25)


ax.set_title("Monthly Sales", fontweight ="bold")
ax.set_xlabel("Sales")
ax.set_ylabel("Month")

fig.tight_layout()
plt.show()
````

## Task 7 — Business analysis

Visualize Sales and Expenses together using a suitable chart.

Then answer:
1. Which month has the highest sales?
2. Which month has the highest expenses?
3. Which month has the highest profit?

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(len(df["Month"]))

width = 0.35

fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(x - width/2, df["Sales"], width, label="Sales", color="skyblue")
ax.bar(x + width/2, df["Expenses"], width, label="Expenses", color="orange")

ax.set_xticks(x)
ax.set_xticklabels(df["Month"])

ax.set_title("Monthly Sales vs Expenses", fontweight="bold")
ax.set_xlabel("Month")
ax.set_ylabel("Money in Rupees")

ax.legend()
ax.grid(True, axis="y", linestyle="--", alpha=0.25)

fig.tight_layout()
plt.show()
````
