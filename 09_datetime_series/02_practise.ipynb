## Task 1 
***Use the same DataFrame and create a chart with these requirements:***
1. Create a figure of size (12, 5).
2. Plot Date against Sales.
3. Use a red line with circle markers.
4. Set title to "Daily Revenue Analysis".
5. Set Y-axis label to "Revenue (₹)".
6. Format dates as 15-Aug using DateFormatter.
7. Use fig.autofmt_xdate().
8. Add a grid.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = pd.DataFrame({
    "Date": [
        "2026-08-01",
        "2026-08-05",
        "2026-08-10",
        "2026-08-15",
        "2026-08-20",
        "2026-08-25"
    ],
    "Sales": [1000, 1500, 1200, 2000, 1800, 2500]
})

df["Date"] = pd.to_datetime(df["Date"])

fig, ax = plt.subplots(figsize = (12, 5))

ax.plot(df["Date"], df['Sales'], marker = 'o', color = "red", markersize = 8)

ax.set_title("Daily Revenue Analysis", fontsize = 12, fontweight= "bold")
ax.set_ylabel("Revenue (₹)", fontsize = 12)
ax.set_xlabel("Date", fontsize=12)

ax.xaxis.set_major_formatter(mdates.DateFormatter("%d-%b"))

ax.grid(True, axis = "both", linestyle = "--", alpha = 0.8)
fig.autofmt_xdate()
fig.tight_layout()
plt.show()
```

    ## Task 2 — Conceptual review

    Q1. What is the purpose of matplotlib.dates?

    Matplotlib.dates provides functions to locate, format and work with dates on Matplotlib axes.


    Q2. What does DateFormatter("%d-%b") display?

    Example:
    01-Aug
    05-Aug
    10-Aug
    15-Aug


    Q3. Why do we use fig.autofmt_xdate()?

    It helps rotate and align date labels so they don't overlap.


    Q4. Difference between plotting month numbers and actual datetime values?

    Here's the important distinction:
    Month numbers	              Actual datetime values
    1, 2, 3, 4	                Jan 5, Jan 15, Feb 10
    Represents month categories	Represents actual dates
    Useful for monthly aggregated data	Useful for daily or irregular time-series data

    Example: Suppose you have sales recorded on January 5 and January 25.

    If you plot only month number 1, both dates have the same X value. But actual datetime values preserve their positions within January.

    That's why actual dates are important for time-series analysis.
