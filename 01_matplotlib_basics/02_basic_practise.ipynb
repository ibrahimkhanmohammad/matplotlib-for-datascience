```console
Create a line graph showing weekly study hours.
    
    X = [1, 2, 3, 4, 5, 6, 7]
    Y = [2, 3, 4, 3, 5, 6, 7]

1. Add the title Weekly Study Hours.
2. Label the X-axis as Days.
3. Label the Y-axis as Study Hours.
4. Display the graph using plt.show().
```

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

X = [1, 2, 3, 4, 5, 6, 7]
Y = [2, 3, 4, 3, 5, 6, 7]

ax.plot(X, Y)

# Add the title Weekly Study Hour
ax.set_title("Weekly Study Hour")

# Label the X-axis as Days.
ax.set_xlabel("Days")

# Label the Y-axis as Study Hours.
ax.set_ylabel("Study Hours")

# Display the graph using plt.show().
plt.show()
````
