## Task 1 — Student marks

    Requirements:
    - Set title, X-axis and Y-axis labels.
    - Use a different bar color.
    - Add black bar borders.
    - Add horizontal grid lines.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

students = ["Ali", "Ahmed", "Sara", "Zoya", "Omar"]
marks = [85, 72, 90, 65, 78]

ax.set_title("Student Marks")

ax.set_xlabel("Student")
ax.set_ylabel("Total Marks")

ax.bar(students, marks, color = "pink", edgecolor = "black")

ax.grid(axis = "y", linestyle = "--")

plt.show()
```

## Task 2 — Monthly revenue

    Requirements:
    - Create a horizontal bar chart.
    - Set appropriate title and axis labels.
    - Change the bar color.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 15000, 10000, 18000, 22000, 20000]

ax.barh(months, revenue, color = "skyblue")

ax.set_title("Monthly Revenue")

ax.set_xlabel("Revenue")
ax.set_ylabel("Month")

ax.grid(axis = "x", linestyle = '--')

plt.show()
```

## Task 3 — Concept questions

    1. Difference between bar() and barh()

    - bar() → vertical bars.
    - barh() → horizontal bars.

    horizontal bars extend along the X-axis; they don't rise from the Y-axis.


    2. What does width control?

    It controls the width of each individual bar.

    For example:
    ax.bar(students, marks, width=0.5)


    3. Why use bar charts for product sales comparisons?

    Products are categories, and bar charts make comparing their values easy.


    4. Bar chart vs Histogram

    Bar chart	            Histogram
    Categorical data	    Numerical data
    Compares categories	    Shows distribution
    Usually has gaps	    Usually has no gaps
    Product sales	        Student marks distribution

    One important detail: a histogram doesn't show spaces between individual records. It groups numerical values into intervals called bins.

    For example, student marks might be grouped into:
    - 0–20
    - 20–40
    - 40–60
    - 60–80
    - 80–100
    The height of each bar tells us how many students fall into that interval.
