## Task 1 — NumPy array visualization

Create two NumPy arrays:
- X = [2, 4, 6, 8, 10]
- Y = [5, 10, 15, 20, 25]
Create a line chart with:
- Green line.
- Circle markers.
- Appropriate title and axis labels.

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([2, 4, 6, 8, 10])
Y = np.array([5, 10, 15, 20, 25])

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(X, Y, marker="o", color="green", markersize=8)

ax.set_title("Numpy array visualization")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")
ax.grid(True, axis="both", linestyle="--", alpha=0.8)

fig.tight_layout()
plt.show()
```

## Task 2 — Using np.arange()
- Generate X values from 0 to 20 with a step of 2.
- Calculate Y values by multiplying X by 3.
- Visualize the result using a line chart.
- Add grid and markers.

```python
X = np.arange(0, 21, 2)
Y = X * 3

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(X, Y, marker="o", color="green", markersize=8)

ax.set_title("Numpy array visualization by arange")
ax.grid(True, axis="both", linestyle="--", alpha=0.8)

fig.tight_layout()
plt.show()
```

## Task 3 — Using np.linspace()
- Generate 50 evenly spaced values between -10 and 10.
- Calculate \(y=x^2\).
- Create a line chart.
- Give it the title "Square Function Visualization".
- Customize the line color and grid.

```python
X = np.linspace(-10, 10, 50)
Y = X**2

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(X, Y, color="red", linewidth = 2)

ax.set_title("Square Function visualization")
ax.grid(True, axis="both", linestyle="--", alpha=0.8)

fig.tight_layout()
plt.show()
```

Q1. Why do we use NumPy arrays with Matplotlib?         
NumPy supports efficient vectorized calculations, so we can perform operations on entire arrays without writing Python loops.

Q2. Difference between np.arange() and np.linspace()?           
np.arange()	np.linspace()       
Uses step size	Uses number of values       
np.arange(0, 10, 2)	np.linspace(0, 10, 5)       
Stop excluded	Endpoint included by default

Q4. Why use 100 values instead of 5 for mathematical functions?         
More points generally make the plotted curve appear smoother.           
With only five points, the curve may look angular. With 100 points, it usually looks smoother.
