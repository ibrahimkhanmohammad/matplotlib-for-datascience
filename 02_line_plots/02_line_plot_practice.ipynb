## Task 1 — Temperature visualization

```console
Requirements:
1. Create a line plot.
2. Set the line color to orange.
3. Set linewidth to 2.
4. Use circle markers.
5. Set marker size to 8.
6. Add title: Weekly Temperature.
7. Label X-axis as Days.
8. Label Y-axis as Temperature (°C).
9. Display the graph.
```

````python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [30, 32, 31, 29, 33, 35, 34]

fig, ax = plt.subplots()

ax.set_title("Weekly Temperature")
ax.set_xlabel("Days")
ax.set_ylabel("Temperature (°C)")

ax.plot(days, temperature, color = "orange", linewidth = 2, marker = 'o', markersize = 8)
plt.show()
````

## Task 2 — Multiple lines

```console
Requirements:

1. Plot both products in one graph.
2. Give Product A a blue line.
3. Give Product B a red line.
4. Add markers to both lines.
5. Give each line a label.
6. Add a legend.
7. Add appropriate title and axis labels.
```

````python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

product_A = [100, 120, 150, 170, 200]
product_B = [80, 110, 130, 160, 190]

fig, ax = plt.subplots()

# Plot both products in one graph, Give Product A a blue line, Give Product B a red line, Add markers to both lines, Give each line a label.
ax.plot(months, product_A, label = "Product A", color = "blue", marker = 'o')
ax.plot(months, product_B, label = "Product B", color = "red", marker = 's')

# Add a legend.
ax.legend()

# Add appropriate title and axis labels.
ax.set_title("Products Sales Comparison")
ax.set_xlabel("Months")
ax.set_ylabel("Sales")

plt.show()
````

```console
Q1. Why do we use line plots?

A more precise Data Science answer:
Line plots are used to visualize trends and changes in data over time or across an ordered sequence. They help identify increases, decreases, patterns and peaks.

For example, monthly sales:
- January → 100
- February → 120
- March → 150
The line helps us understand how sales are increasing.


Q2. Difference between linewidth and markersize?

Parameter	Meaning
linewidth	Controls the thickness of the connecting line
markersize	Controls the size of individual data point markers

Q3. Purpose of ax.legend()?


ax.legend() displays the labels assigned to plotted elements.

For example:
ax.plot(months, product_A, label="Product A")ax.plot(months, product_B, label="Product B")ax.legend()
The legend helps us identify which line represents Product A and which represents Product B.

Q4. What happens without ax.legend()?

The lines will still be plotted, but their labels won't appear on the graph.
This can make the visualization confusing when multiple lines are present.
```
