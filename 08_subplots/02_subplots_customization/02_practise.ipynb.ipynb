## Task 1 — Monthly revenue visualization

Use this dataset:
```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 15000, 10000, 18000, 22000, 20000]
```

Requirements:
1. Create a figure with figsize=(10, 5).
2. Create a line plot with markers.
3. Customize the title with fontsize=16 and fontweight="bold".
4. Add X-axis and Y-axis labels.
5. Set Y-axis limits from 0 to 25000.
6. Add horizontal dashed grid lines.
7. Highlight the highest revenue using annotate().
8. Display a legend.

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize = (10, 5))

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 15000, 10000, 18000, 22000, 20000]

ax.plot(months, revenue, marker = 'o', markersize = 8, label = "Monthly Sales")
ax.set_title("Monthly Revenue", fontsize = 16, fontweight = "bold")
ax.set_xlabel("Months", fontsize = 12)
ax.set_ylabel("Revenue", fontsize = 12)
ax.set_ylim(0, 25000)
ax.grid(True, axis = "y", linestyle = "--", alpha = 0.5)
ax.annotate("Highest Revenue", xy = ("May", 22000), xytext = ("Feb", 23000), arrowprops = dict(arrowstyle = "->", color = "red"), fontsize = 10)
ax.legend(loc = "best")

fig.tight_layout()

plt.show()
````

## Task 2 — Product comparison

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

product_A = [100, 120, 150, 170, 200]
product_B = [80, 110, 130, 160, 190]
```

Requirements:
1. Create a figure with figsize=(10, 5).
2. Plot both products on the same Axes.
3. Customize their colors and markers.
4. Add a title and axis labels.
5. Customize the grid.
6. Display a legend in the upper-left corner.
7. Customize tick-label font size.

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

product_A = [100, 120, 150, 170, 200]
product_B = [80, 110, 130, 160, 190]

fig, ax = plt.subplots(figsize = (10, 5))

ax.plot(months, product_A, label = "Product A", marker = 'o', color = "green", markersize = 6)
ax.plot(months, product_B, label = "Product B", marker = 's', color = "red", markersize = 6)

ax.set_title("Monthly Product Comparison", fontsize = 12, fontweight = "bold")
ax.set_xlabel("Month", fontsize = 12)
ax.set_ylabel("Total Products", fontsize=12)
ax.grid(True, axis = "y", linestyle = "--", alpha = 0.5)
ax.legend(loc = "upper left", fontsize = 12)
ax.tick_params(
    axis="both",
    labelsize=10,
    labelcolor="magenta"
)
ax.set_ylim(0, 300)

fig.tight_layout()

plt.show()
````

## Task 3 - Conceptual Questions

1. figsize

Controls the width and height of the entire figure in inches.

2. set_xlim() and set_ylim()
- set_xlim() controls the displayed X-axis range.
- set_ylim() controls the displayed Y-axis range.
3. annotate()

Adds explanatory text and optionally an arrow to highlight a particular point or observation.

4. xy vs xytext
- xy → location of the actual data point being annotated.
- xytext → location where annotation text appears.
5. alpha

Controls transparency.
- 1.0 → fully opaque.
- 0.5 → partially transparent.
- 0.0 → fully transparent.
6. Legends

Help identify which plotted line or data series corresponds to each label.
