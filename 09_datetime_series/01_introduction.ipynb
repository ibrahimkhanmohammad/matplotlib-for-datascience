## Introduction
In real-world Data Science, we frequently work with dates and time-based data.

For example:
- Daily stock prices.
- Monthly company revenue.
- Website visitors over time.
- Daily temperature records.
- Monthly product sales.
A time-series chart helps us understand how values change over time.

## Important functions
We'll learn two functions.
A. mdates.DateFormatter()
It controls how dates appear on the X-axis.

Syntax:
```python
mdates.DateFormatter("%d-%b")
```
        Common formatting codes:
        Code	Meaning	           Example
        %d	    Day	                    15
        %m	    Month number	    08
        %b	    Short month name        Aug
        %B	    Full month name	    August
        %Y	    Full year	            2026
        %y	    Short year	            26

Examples:

```python
"%d-%b"       # 15-Aug
"%b-%Y"       # Aug-2026
"%d-%m-%Y"    # 15-08-2026
"%B %Y"       # August 2026
```

## fig.autofmt_xdate()
When date labels overlap, this function automatically rotates and adjusts them to improve readability.

Syntax:
```python
fig.autofmt_xdate()
```

## Practical example
Let's create a daily sales time-series chart.


````python
# Step 1 — Create the dataset

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = pd.DataFrame({
    "Date": [
        "2026-08-01",
        "2026-08-05",
        "2026-08-10",
        "2026-08-15",
        "2026-08-20",
        "2026-08-25"
    ],
    "Sales": [1000, 1500, 1200, 2000, 1800, 2500]
})

df["Date"] = pd.to_datetime(df["Date"])
````

````python
# Step 2 — Create the chart

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(df["Date"], df["Sales"],
        marker="o", color="green")

ax.set_title("Daily Sales Trend")
ax.set_xlabel("Date")
ax.set_ylabel("Sales (₹)")

ax.xaxis.set_major_formatter(
    mdates.DateFormatter("%d-%b")
)

fig.autofmt_xdate()

ax.grid(True, linestyle="--", alpha=0.5)

plt.show()
````

## Understand the new code
        Code	                                Purpose
        import matplotlib.dates as mdates	    Imports date-formatting tools
        df["Date"]                              Provides actual dates on X-axis
        ax.xaxis                                Accesses the X-axis
        set_major_formatter()	            Changes how major date ticks are displayed
        DateFormatter("%d-%b")	            Displays dates like 01-Aug
        fig.autofmt_xdate()             	    Adjusts date-label rotation and spacing

Important distinction:

Previously, we plotted month numbers (1, 2, 3). Now we're plotting actual datetime values directly.

##  When should you use this?
Use date formatting when:
- Your X-axis contains dates.
- Date labels overlap.
- You want to display days, months or years clearly.
- You're visualizing daily, weekly or monthly trends.
