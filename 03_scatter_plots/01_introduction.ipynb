```console
A scatter plot is used to visualize the relationship between two numerical variables.
Unlike a line plot, a scatter plot displays individual data points without connecting them.

For example, imagine we have student data:

Study Hours	 Exam Marks
    1	            35
    2	            40
    3	            50
    4	            60
    5	            75
    6	            85
```

## Creating your first scatter plot

````python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
exam_marks = [35, 40, 50, 60, 75, 85]

fig, ax = plt.subplots()

ax.scatter(study_hours, exam_marks)

ax.set_title("Study Hours vs Exam Marks")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Exam Marks")

plt.show()
````

## Customizing scatter plots

### Changing color

````python
fig, ax = plt.subplots()

ax.scatter(study_hours, exam_marks, color = 'green')
plt.show()
````

### Changing marker size

```console
Unlike we use markersize in line plots here we use s

Why s?
 >> It controls the marker area in points squared, rather than the marker's linear size.
```

````python
fig, ax = plt.subplots()

ax.scatter(study_hours, exam_marks, color = 'green', s=20)
plt.show()
````

### Changing marker shape

````python
fig, ax = plt.subplots()

ax.scatter(study_hours, exam_marks, color = 'green', s=20, marker = 'x')
plt.show()
````

### Adding transparency

    This is a particularly useful feature when working with large datasets

    alpha=1 → Fully opaque.
    alpha=0.5 → Semi-transparent.
    alpha=0 → Fully transparent.

````python
fig, ax = plt.subplots()

ax.scatter(study_hours, exam_marks, color = 'green', s=20, marker = 'x', alpha = 0.5)
plt.show()
````

## Understanding relationships in scatter plots

```console
Scatter plots help us identify three common relationships.

A. Positive correlation
When X increases, Y generally increases.
    > Example: Study hours and exam marks.

B. Negative correlation
When X increases, Y generally decreases.
    > Example: Product price and quantity demanded (often)

C. No clear correlation
    > There is no obvious linear relationship between the two variables.

Important: Correlation does not automatically mean causation. A scatter plot helps us investigate relationships, not prove that one variable causes another.
```
