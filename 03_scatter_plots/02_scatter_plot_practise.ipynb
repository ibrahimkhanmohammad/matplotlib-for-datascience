## Task 1 — Study hours vs Exam marks

    Requirements:

    1. Create a scatter plot.
    2. Set color to purple.
    3. Set marker size to 100.
    4. Use circle markers.
    5. Set alpha to 0.7.
    6. Add an appropriate title.
    7. Label both axes.
    8. Display the plot.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
exam_marks = [30, 35, 45, 50, 60, 70, 80, 90]

# Create a scatter plot, Set color to purple, Set marker size to 100, Use circle markers, Set alpha to 0.7. 
ax.scatter(study_hours, exam_marks, color = "purple", s=100, marker = 'o', alpha = 0.7)

# Add an appropriate title.
ax.set_title("Study Hours vs Exam Marks")

# Label both axes.
ax.set_xlabel("Study Hours")
ax.set_ylabel("Total Marks")

# Display the plot.
plt.show()
```

## Task 2 — Advertising vs Sales

    Requirements:

    1. Create a scatter plot.
    2. Set the marker color to green.
    3. Set marker size to 80.
    4. Use triangle markers.
    5. Add title: Advertising vs Sales.
    6. Label X-axis as Advertising Expenditure.
    7. Label Y-axis as Sales.
    8. Add grid lines.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

advertising = [10, 20, 30, 40, 50, 60, 70, 80]
sales = [15, 25, 28, 40, 45, 55, 65, 75]

# Create a scatter plot, Set marker size to 80, Set the marker color to green, Use triangle markers
ax.scatter(advertising, sales, s=80, color = "green", marker = '^')

# Add title: Advertising vs Sales.
ax.set_title("Advertising vs Sales")

# Label X-axis as Advertising Expenditure.
ax.set_xlabel("Advertising Expenditure")

# Label Y-axis as Sales.
ax.set_ylabel("Sales")

# Add grid lines.

ax.grid(True)

# Display the Plot
plt.show()
```

## Task 3 — Conceptual evaluation

    Q1. What is a scatter plot?

    A scatter plot is a visualization that represents the relationship between two numerical variables using individual points.

    Q2. What does each dot represent?

    Each dot represents one observation (one record) containing two values.

    For example:
    Student	    Study hours (X)	    Marks (Y)
    A	        2	                40
    B	        4	                60
    C	        6	                85

    Each student produces one dot:
    - Student A → (2, 40)
    - Student B → (4, 60)
    - Student C → (6, 85)

    So remember: One dot = One observation with two numerical values.

    Q3. Difference between markersize and s

    Parameter	    Used in	Meaning
    markersize	    ax.plot()	Marker size in points
    s	            ax.scatter()	Marker area in points squared

    One technical correction: s is not exactly the same measurement as markersize. It controls the marker's area, not its linear size.

    Q4. What does alpha control?

    alpha controls transparency.
    Value	Appearance
    1.0	Fully opaque
    0.7	Slightly transparent
    0.5	Semi-transparent
    0.1	Highly transparent
    0.0	Invisible

    Q5. What is positive correlation?

    When X increases, Y generally increases.

    Q6. What is negative correlation?

    When X increases, Y generally decreases.

    Q7. Can correlation prove causation?

    This is one of the most important concepts in Data Science.

    Let's understand with an example.

    Imagine we observe:
    - Ice cream sales increase during summer.
    - Swimming pool accidents also increase during summer.
    Does buying ice cream cause swimming pool accidents?

    Of course not!

    Both may increase because of a third factor: warmer weather.

    This illustrates:
    - Correlation: Two variables tend to change together.
    - Causation: A change in one variable directly produces a change in another.
    Remember this sentence:
    Correlation does not imply causation.
