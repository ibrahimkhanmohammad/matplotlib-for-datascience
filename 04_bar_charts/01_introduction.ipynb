    Bar charts are one of the most useful visualizations in Data Science. We use them to compare values across different categories.

    What is a bar chart?
    Imagine we have sales data for three products:
    
    Product	    Sales
    Laptop	    500
    Mobile	    800
    Tablet	    300

    We want to compare their sales. A bar chart makes that comparison easy.

## Creating a basic bar chart

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet"]
sales = [500, 800, 300]

fig, ax = plt.subplots()

ax.bar(products, sales)

ax.set_title("Product Sales")
ax.set_xlabel("Products")
ax.set_ylabel("Sales")

plt.show()

'''
Understand the output
- X-axis contains product categories.
- Y-axis contains sales values.
- Each bar represents one product.
- Taller bar means higher sales.
'''
```

## Customizing bar charts

    Parameter	    Purpose
    color	        Changes bar color
    width	        Changes bar width
    edgecolor       Changes border color
    alpha	        Controls transparency
    align	        Controls bar alignment

    Note: width is generally between 0 and 1 for categorical bar positions.

```python
fig, ax = plt.subplots()

ax.bar(
    products,
    sales,
    color="skyblue",
    width=0.5,
    edgecolor="black",
    alpha=0.8,
    align='edge'
)

ax.set_title("Product Sales")
ax.set_xlabel("Products")
ax.set_ylabel("Sales")

ax.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

## Horizontal bar chart
***Sometimes category names are long. Horizontal bars make them easier to read.***

    Remember the difference:
    - bar() → vertical bars.
    - barh() → horizontal bars.

```python
fig, ax = plt.subplots()

ax.barh(products, sales, color="orange")

ax.set_title("Product Sales")
ax.set_xlabel("Sales")
ax.set_ylabel("Products")

plt.show()
```

    Bar chart vs Histogram
    This is very important for Data Science.
    
    Bar chart	                Histogram
    Compares categories	        Shows numerical distribution
    Bars represent categories	Bars represent numerical intervals (bins)
    Usually has gaps	        Usually bars touch
    Example: Sales by product	Example: Distribution of student marks
