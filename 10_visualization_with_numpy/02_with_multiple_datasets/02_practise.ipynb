## Task 1 — Compare three functions

Generate 100 evenly spaced X values between -5 and 5.

Create these three datasets:
- Y1 = X
- Y2 = X ** 2
- Y3 = X ** 3
Requirements:
- Plot all three on the same axes.
- Use three different colors.
- Give each line a label.
- Add a legend.
- Add title, X-axis label and Y-axis label.
- Add a grid.

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.linspace(-5, 5, 100)
Y1 = X
Y2 = X**2
Y3 = X**3

fig, ax = plt.subplots(figsize=(12, 7))

ax.plot(X, Y1, label="Y1 = X", color="skyblue")
ax.plot(X, Y2, label="Y2 = X ** 2", color="pink")
ax.plot(X, Y3, label="Y3 = X ** 3", color="orange")

ax.set_title("Comparison of Three Mathematical Functions")
ax.set_xlabel("x-values")
ax.set_ylabel("y-values")
ax.grid(True, alpha=0.25)
ax.legend(loc="best")

fig.tight_layout()
plt.show()
```

## Task 2 — Compare two datasets

Generate X values from 0 to 10 using np.arange() with a step of 1.

Calculate:
- Y1 = X * 2
- Y2 = X * 5
Plot both datasets on the same chart.

Requirements:
- Different line colors.
- Different markers.
- Legend.
- Appropriate title and labels.

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.arange(0, 11)
Y1 = X * 2
Y2 = X * 5


fig, ax = plt.subplots(figsize=(12, 7))

ax.plot(X, Y1, label="Y1 = X * 2", color="skyblue", marker = 'o')
ax.plot(X, Y2, label="Y2 = X * 5", color="pink", marker = 's')


ax.set_title("Comparison of Two Mathematical Functions")
ax.set_xlabel("x-values")
ax.set_ylabel("y-values")
ax.grid(True, alpha=0.25)
ax.legend(loc="best")

fig.tight_layout()
plt.show()
```

## Conceptual questions — Evaluation

Q1. How can we plot multiple datasets on the same axes?             
Use multiple ax.plot() calls with the same Axes object.     

Q2. Why do we use label and ax.legend()?            
- label assigns a name to each plotted dataset.         
- ax.legend() displays those names in a legend.         

Q3. What happens if we create three separate ax.plot() calls using the same ax?     
Three separate plots (lines) appear on the same chart.      

They don't create three different charts.       

Q4. Why is comparing multiple datasets on one chart useful?     
It makes comparisons easier without switching between multiple charts.
