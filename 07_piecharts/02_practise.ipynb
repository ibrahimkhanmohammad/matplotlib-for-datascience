## Task 1 — Company expenses

Create a pie chart using:
       
categories = ["Rent", "Salary", "Marketing", "Utilities", "Other"]      
expenses = [30000, 50000, 10000, 5000, 5000]

Requirements:
1. Display category labels.
2. Display percentages with one decimal place.
3. Give each slice a different color.
4. Set the title to "Company Expenses Distribution".

```python
import matplotlib.pyplot as plt

categories = ["Rent", "Salary", "Marketing", "Utilities", "Other"]
expenses = [30000, 50000, 10000, 5000, 5000]

fig, ax = plt.subplots()

colors = ["skyblue", "pink", "lightgreen", "orange", "magenta"]
ax.pie(expenses, labels = categories, autopct = "%1.1f%%", colors = colors)

ax.set_title("Company Expenses Distribution")
plt.show()
```

## Task 2 — Smartphone market share

use:        
brands = ["Samsung", "Apple", "Xiaomi", "Vivo", "Others"]           
market_share = [30, 25, 20, 15, 10]

Requirements:
1. Display labels and percentages.
2. Set startangle=90.
3. Explode the Apple slice.
4. Add a suitable title.

```python
import matplotlib.pyplot as plt

brands = ["Samsung", "Apple", "Xiaomi", "Vivo", "Others"]
market_share = [30, 25, 20, 15, 10]

fig, ax = plt.subplots()

explode = [0, 0.1, 0, 0, 0]
colors = ["skyblue", "pink", "lightgreen", "orange", "magenta"]

ax.pie(market_share, labels = brands, autopct = "%1.1f%%", colors = colors, startangle = 90, explode = explode)

ax.set_title("Smartphone Market Share Distribution")
plt.show()
```

## Task 3 - Quick conceptual check

1. What does autopct do?            
autopct: Automatically displays each slice's percentage.

2. What is the purpose of explode?          
explode: Separates selected slices from the centre of the pie.

3. What does startangle control?            
startangle: Sets the starting rotation angle of the pie chart.

4. Why should we avoid pie charts with too many categories?         
Too many categories: Makes slices difficult to distinguish and compare.

5. If you want to compare sales of 10 products, would you choose a pie chart or bar chart? Why?         
Ten products: Bar chart is better because comparing bar lengths is easier than comparing many pie slices.
