# Customization, Annotations & Styling
***We'll learn how to make Matplotlib charts look professional and suitable for real-world Data Science project***

## Figure size — figsize
Until now, we've been creating figures like this:

fig, ax = plt.subplots(figsize=(10, 5))

* 10 → width in inches.
* 5 → height in inches.

**Remember: figsize controls the overall figure dimensions, not the size of individual data points.**

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(months, sales, marker="o")

ax.set_title("Monthly Sales")
ax.set_xlabel("Months")
ax.set_ylabel("Sales")

plt.show()
````

## Grid customization

```python
ax.grid(
    True,
    axis="y",
    linestyle="--",
    linewidth=0.7,
    alpha=0.7
)
```

        Understand the parameters:
        Parameter	        Purpose
        True	        Displays grid
        axis="y"	        Shows horizontal grid lines
        axis="x"	        Shows vertical grid lines
        linestyle="--"	Dashed lines
        linewidth=0.7	Grid line thickness
        alpha=0.7	        Transparency

Other line styles:
- "-" → solid
- "--" → dashed
- ":" → dotted
- "-." → dash-dot

## Font customization
We can customize chart titles, axis labels and tick labels.

### A. Title customization

```python
ax.set_title("Monthly Sales", fontsize=18, color="blue", fontweight="bold")
```

### B. Axis labels
```python
ax.set_xlabel("Months", fontsize=12)
ax.set_ylabel("Revenue", fontsize=12)
```

### C. Tick labels
Tick labels are the values displayed along the X and Y axes.
```python
ax.tick_params(axis="both", labelsize=10, colors="green")
```

* axis="x" → X-axis tick labels.
* axis="y" → Y-axis tick labels.
* axis="both" → both axes.

## Axis limits — set_xlim() and set_ylim()
Sometimes Matplotlib automatically selects the axis range.

But we may want to control it ourselves.

```python
ax.set_xlim(0, 6)
ax.set_ylim(0, 250)
```

* set_xlim() → controls the X-axis range.
* set_ylim() → controls the Y-axis range.

````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [100, 150, 120, 180, 200]

fig, ax = plt.subplots()

ax.plot(months, sales, marker="o")

ax.set_xlim(0, 6)
ax.set_ylim(0, 250)

plt.show()
````

## Annotations — ax.annotate()
Important for EDA
Annotations help us highlight a particular point or observation in a chart.

For example, imagine monthly sales are increasing, and April has the highest sales.

We can highlight that point with a message and an arrow.
Syntax:
```python

ax.annotate(
    "Highest Sales",
    xy=(4, 200),
    xytext=(2.5, 230),
    arrowprops=dict(arrowstyle="->")

)

```

`Understand the arguments:
* xy → actual data point we want to highlight.
* xytext → position where the annotation text should appear.
* arrowprops → controls the arrow appearance.
Let's understand with complete code:
``

````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [100, 150, 120, 180, 200]

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(months, sales, marker="o")

ax.annotate(
    "Highest Sales",
    xy=(5, 200),
    xytext=(3, 230),
    arrowprops=dict(arrowstyle="->", color="red"),
    fontsize=12
)

ax.set_title("Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

plt.show()
````

## Legends customization

Now let's customize it.
```python
ax.legend(
    loc="upper left",
    fontsize=10,
    title="Products"
)
```
Common legend positions:
```console
        Value	        Position
        "upper right"	Top right
        "upper left"	Top left
        "lower right"	Bottom right
        "lower left"	Bottom left
        "best"	        Automatically selected
   ```

Example with multiple lines:


````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

product_A = [100, 120, 150, 180]
product_B = [80, 110, 130, 160]

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(months, product_A, label="Product A", marker="o")
ax.plot(months, product_B, label="Product B", marker="s")

ax.legend(loc="best", fontsize=10, title="Products")

plt.show()
````

## Complete professional visualization
Let's combine what we've learned.

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    months,
    sales,
    color="blue",
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

ax.set_title(
    "Monthly Sales Analysis",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel("Months", fontsize=12)
ax.set_ylabel("Sales", fontsize=12)

ax.set_ylim(0, 250)

ax.grid(True, axis="y", linestyle="--", alpha=0.5)

ax.annotate(
    "Highest Sales",
    xy=("May", 200),
    xytext=("Mar", 230),
    arrowprops=dict(arrowstyle="->", color="red")
)

ax.legend(loc="upper left")

fig.tight_layout()

plt.show()
````
